Q7. Write a python program to find the longest words.
#steps- 
# 1- Open the file (no need to print anything now) 
# 2- Store the content in a variable
# 3- Now split the content (sentence into words) using a new variable
# 4- Once split done then create a variable to store the longest word
# 5- Using for loop and len function compare each word.



```python

with open("file3.txt","r") as file:
    content = file.read()
    
words= content.split()

longest_word= ""

for i in words:
    if len(i) > len(longest_word):
        longest_word=i
        
print(f'longest word is "{longest_word}" and its length is {len(longest_word)}')
        

    
    
```

Q9-Write a Python program to count the frequency of words in a file.


```python

with open("file4.txt","r") as file:
    content=file.read()
    
words=content.split()

word_count={}

for i in words:
    if i in word_count:
        word_count[i]= word_count[i]+1
    else:
        word_count[i]=1
        
print(word_count)
```
