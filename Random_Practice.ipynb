```python
#star pattern- increasing pattern

n=6

for rows in range(n):
  for cols in range(rows+1):
    print("*",end=" ")
  print()
```

```python
#star pattern- Decreasing pattern

n=10

for rows in range(n):

  for cols in range(rows,n):
    print("",end=" ")   #-> here its not ending and gng on second line, after printing the spaces it would go on below loop

  for cols in range(rows+1):
    print("*",end=" ")

  print()
```
