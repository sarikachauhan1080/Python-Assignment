```python
#find the duplicate characters in the string..

text = "programming"

duplicate={}

for i in text:
    if i in duplicate:
        duplicate[i] +=1
    else:
        duplicate[i] = 1
          

for i in duplicate:
    if duplicate[i] > 1:
        print(i)
```

```python
#find duplicate numbers:


numbers = [1, 2, 3, 2, 4, 1, 5, 2]

repeat={}

for i in numbers:
    if i in repeat:
        repeat[i] += 1
    else:
        repeat[i]=1

for i in repeat:
    if repeat[i]>1:
        print(i)
```

```python
#inverted star

for i in range(1,6):
    print(" " * (i-1) + "*" * (2*(5-i)-1))
```

```python
#right aligned triangle

for i in range(1,6):
    print(" " * (6-i) + "*" * i)
```

```python
#inverted right angle triangle

for i in range(1,6):
    print(" " * (i-1) + "*" * (6-i) )
```

```python
#Diamond

for i in range(1,6):
    print(" " * (6-i) + "*" * (2*i-1))
for i in range(2,6):
    print(" " * (i) + "*" * (2*(5-i)+1))
```

```python
#hourglass

for i in range(1,6):
    print(" " * (i) + "*" * (2*(5-i)+1))
for i in range(1,6):
    print(" " * (5-i) + "*" * (2*i+1))
```

```python
def sum(n):
    if n == 0:
        return 0
    
    return n + sum(n-1)

print(sum(5))
```

```python
class Employee:
    
    def __init__(self, name, salary, department):
        self.name=name
        self.salary=salary
        self.department=department
             
    def display_details(self):
        print(self.name)
        print(self.salary)
        print(self.department)


Sarika=Employee("Sarika",100000,"IT")
Yashpal=Employee("Yashpal",100000,"Infra")

Sarika.display_details()
Yashpal.display_details()

    
```

```python
class Father:
    
    def House(self):
        print("Father has a House")
        
    def Car(self):
        print("Father has a car")
        
class Son(Father):
    
    def Bike(self):
        print("Son has a Bike")
        
Rahul =Son()

Rahul.House()
Rahul.Bike()
Rahul.Car()
```

```python
class Employee:

        def __init__(self,name,salary):
            self.name=name
            self.salary=salary
            
        def display_employee(self):
            print(self.name,self.salary)
            
class Manager(Employee):
    
    def __init__(self, name, salary, department):
         super().__init__(name,salary) #this is the parent constructor, it only accepts name and salary so only pass those two...
         self.department=department
         
    def display_manager(self):
         print(self.name,self.department,self.salary)
         
         
manager1=Manager("Sarika","IT",100000)

manager1.display_manager()
         
         
            
```
